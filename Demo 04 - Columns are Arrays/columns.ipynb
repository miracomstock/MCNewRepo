## Practice

Find the total salary paid to Chicago Bulls players in the year 2010.
- Use where to narrow the table to the right team and year.
- Use column to extract the correct column as an array.
- Use the sum command on the array.

```python
bulls = nba.where('team', 'Chicago Bulls')
bulls2010 = bulls.where('season', 2010)
#bulls2010.show()
salary_arr = bulls2010.column('salary')
np.average(salary_arr)

```

## with_column

```python
nba = Table.read_table('nba_salaries.csv')
warriors = nba.where('team', 'Golden State Warriors')
warriors.show(10)
```

```python
salary_dollars = warriors.column('salary')
salary_euros = np.round(0.86 * salary_dollars)
salary_euros
```

```python
# Attach new column to table
warriors = warriors.with_column('salary euros', salary_euros)
warriors.show(10) # salaries turned from into to float, but that's okay
```
